```python
##fib
import time
def fib(n):
  if n == 1 or n == 0:
    return 1
  else:
    return fib(n-1) + fib (n-2)
```

```python
fib(5)
```

```python
fib(36)
```

```python
## it takes time
```

```python
start = time.time()
fib(38)
print(time.time() - start)
```

```python
##solution to reduce time  means store that results that repeats multiple times
#update the code and add Dictinary
import time
def fib(n ,d):
  if n in d:
    return d[n]
  else:
    d[n] =fib(n-1,d) + fib (n-2,d)
    return d[n]
```

```python
start = time.time()
d = {0:1,1:1}
fib(38 ,d)
print(time.time() - start)
```
